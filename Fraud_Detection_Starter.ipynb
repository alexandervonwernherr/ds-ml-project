## Feature Engineering

```python
#convert the column invoice_date to date time format on both the invoice train and invoice test
for df in [invoice_train,invoice_test]:
    df['invoice_date'] = pd.to_datetime(df['invoice_date'])
print("Train:", invoice_train["invoice_date"].min(), "to", invoice_train["invoice_date"].max())
print("Test: ", invoice_test["invoice_date"].min(), "to", invoice_test["invoice_date"].max())
```

```python
#encode labels in categorical column
d={"ELEC":0,"GAZ":1}
invoice_train['counter_type']=invoice_train['counter_type'].map(d)
invoice_test['counter_type']=invoice_test['counter_type'].map(d)
```

```python
#convert categorical columns to int for model
client_train['client_catg'] = client_train['client_catg'].astype(int)
client_train['disrict'] = client_train['disrict'].astype(int)

client_test['client_catg'] = client_test['client_catg'].astype(int)
client_test['disrict'] = client_test['disrict'].astype(int)
```

```python
def aggregate_by_client_id(invoice_data):
    aggs = {}
    aggs['consommation_level_1'] = ['mean']
    aggs['consommation_level_2'] = ['mean']
    aggs['consommation_level_3'] = ['mean']
    aggs['consommation_level_4'] = ['mean']

    agg_trans = invoice_data.groupby(['client_id']).agg(aggs)
    agg_trans.columns = ['_'.join(col).strip() for col in agg_trans.columns.values]
    agg_trans.reset_index(inplace=True)

    df = (invoice_data.groupby('client_id')
            .size()
            .reset_index(name='{}transactions_count'.format('1')))
    return pd.merge(df, agg_trans, on='client_id', how='left')
```

```python
#group invoice data by client_id
agg_train = aggregate_by_client_id(invoice_train)
```

```python
print(agg_train.shape)
agg_train.head()
```

```python
#merge aggregate data with client dataset
train = pd.merge(client_train,agg_train, on='client_id', how='left')
```

```python
#aggregate test set
agg_test = aggregate_by_client_id(invoice_test)
test = pd.merge(client_test,agg_test, on='client_id', how='left')
```

```python
train.shape, test.shape
```

```python
# 1. Invoices per client (already in agg_train)
print(agg_train["1transactions_count"].describe())

# 2. Counter type per client: ELEC only, GAZ only, or both
# (counter_type was encoded earlier: ELEC=0, GAZ=1)
types = invoice_train.groupby("client_id")["counter_type"].agg(["min", "max"])
types["counter_mix"] = types.apply(
    lambda r: "both" if r["min"] != r["max"] else ("ELEC only" if r["min"] == 0 else "GAZ only"),
    axis=1,
)

eda = agg_train.merge(types["counter_mix"], on="client_id")

# 3. Invoice count by counter type
print(eda["counter_mix"].value_counts())
print(eda.groupby("counter_mix")["1transactions_count"].describe())
```

```python
#drop redundant columns
sub_client_id = test['client_id']
drop_columns = ['client_id', 'creation_date']

for col in drop_columns:
    if col in train.columns:
        train.drop([col], axis=1, inplace=True)
    if col in test.columns:
        test.drop([col], axis=1, inplace=True)
```

# Modelling

## Train LGBM *Classifier*

```python
x_train = train.drop(columns=['target'])
y_train = train['target']
```

```python
from sklearn.model_selection import train_test_split

x_tr, x_val, y_tr, y_val = train_test_split(
    x_train, y_train, test_size=0.2, stratify=y_train, random_state=42
)
```

```python
model = LGBMClassifier(boosting_type='gbdt', num_iterationss=500)
#model.fit(x_train, y_train)
model.fit(x_tr, y_tr)
```

```python
x_train.disrict.unique()
```

## Make Predictions on test set

```python
test.columns
```

```python
preds = model.predict(test)
preds = pd.DataFrame(preds, columns=['target'])
preds.head()
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

val_scores = model.predict_proba(x_val)[:, 1]
auc = roc_auc_score(y_val, val_scores)

fpr, tpr, _ = roc_curve(y_val, val_scores)
plt.plot(fpr, tpr, label=f"Model (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], "--", color="grey", label="Baseline (AUC = 0.5)")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.legend()
plt.show()
```

```python
#trying logistic regression
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

logreg = make_pipeline(
    StandardScaler(),
    LogisticRegression(class_weight="balanced", max_iter=1000),
)
logreg.fit(x_tr, y_tr)

lr_scores = logreg.predict_proba(x_val)[:, 1]
lr_auc = roc_auc_score(y_val, lr_scores)

lgbm_fpr, lgbm_tpr, _ = roc_curve(y_val, val_scores)
lr_fpr, lr_tpr, _ = roc_curve(y_val, lr_scores)

plt.plot(lgbm_fpr, lgbm_tpr, label=f"LightGBM (AUC = {auc:.3f})")
plt.plot(lr_fpr, lr_tpr, label=f"Logistic regression (AUC = {lr_auc:.3f})")
plt.plot([0, 1], [0, 1], "--", color="grey", label="Baseline (AUC = 0.5)")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve comparison - validation set")
plt.legend()
plt.show()
```

```python
neg, pos = (y_tr == 0).sum(), (y_tr == 1).sum()
print(f"scale_pos_weight = {neg / pos:.1f}")

model_bal = LGBMClassifier(boosting_type="gbdt", n_estimators=500, scale_pos_weight=neg / pos)
model_bal.fit(x_tr, y_tr)

bal_scores = model_bal.predict_proba(x_val)[:, 1]
bal_auc = roc_auc_score(y_val, bal_scores)

print(f"Original LightGBM AUC: {auc:.3f}")
print(f"Balanced LightGBM AUC: {bal_auc:.3f}")

bal_fpr, bal_tpr, _ = roc_curve(y_val, bal_scores)
plt.plot(lgbm_fpr, lgbm_tpr, label=f"LightGBM (AUC = {auc:.3f})")
plt.plot(bal_fpr, bal_tpr, label=f"LightGBM balanced (AUC = {bal_auc:.3f})")
plt.plot([0, 1], [0, 1], "--", color="grey", label="Baseline (AUC = 0.5)")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Effect of fixing the imbalance - validation set")
plt.legend()
plt.show()
```
